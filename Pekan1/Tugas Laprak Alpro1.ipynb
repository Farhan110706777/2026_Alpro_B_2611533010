Pemrograman komputer merupakan fondasi utama dalam pengembangan perangkat lunak modern. ## 2.1 Dasar Teori
Dalam praktikum modul pertama ini, terdapat 3 komponen utama yang dipelajari:

1. Bahasa Pemrograman Python:
   Python adalah bahasa pemrograman tingkat tinggi (high-level language) bersifat interpreted dan berorientasi objek (object-oriented). Python diciptakan oleh Guido van Rossum dan dirilis pertama kali pada tahun 1991. Python menekankan pada keterbacaan kode (code readability) dengan memanfaatkan indentasi secara konsisten sebagai pengganti tanda kurung kurawal.
 2. Visual Studio Code (VS Code):
   Visual Studio Code adalah code editor sumber terbuka (open-source) yang dikembangkan oleh Microsoft. VS Code mendukung berbagai bahasa pemrograman melalui sistem extension. Fitur-fitur unggulannya meliputi IntelliSense (auto-pilihan dan sintaksis highlighting), integrated terminal, debugger, serta integrasi bawaan (built-in) dengan sistem Git.
 3. Git dan GitHub:
   * Git adalah sistem kontrol versi terdistribusi (Distributed Version Control System / DVCS) yang bekerja di tingkat lokal komputer pengembang untuk mencatat riwayat perubahan berkas berkode.
   * GitHub adalah platform berbasis cloud yang menyediakan layanan hosting repositori Git. GitHub memungkinkan pengembang menyimpan cadangan kode secara daring, berkolaborasi secara tim, serta mempublikasikan repositori proyek secara terbuka maupun privat.
 4. Koneksi dan Alur Kerja Git/GitHub di VS Code:
   Alur kerja umum menghubungkan repositori lokal dengan GitHub mencakup perintah-perintah dasar berikut:
   a. git init: Menginisialisasi repositori Git baru di folder lokal.
   b. git add: Menambahkan perubahan berkas ke dalam staging area.
   c. git commit -m "pesan": Menyimpan snapshot perubahan berkas ke riwayat lokal beserta deskripsi pesan.
   d. git remote add origin <URL>: Menghubungkan repositori lokal dengan repositori remote di GitHub.
   e. git push -u origin main: Mengunggah (upload) seluruh commit lokal ke branch utama di GitHub.

## 2.1 Dasar Teori
Struktur kontrol percabangan di Python digunakan untuk menentukan blok kode mana yang akan dieksekusi berdasarkan evaluasi ekspresi *boolean* (Benar/Salah). Terdapat beberapa bentuk arsitektur percabangan dasar:

1. **If Tunggal**: Struktur paling sederhana yang hanya akan mengeksekusi blok kode di bawahnya jika kondisi bernilai `True`.
2. **If-Else**: Memberikan alternatif jalan (blok kode lain) apabila kondisi utama bernilai `False`.
3. **If-Elif-Else**: Digunakan untuk mengevaluasi lebih dari dua kemungkinan. Kondisi dievaluasi secara berurutan (sekuensial) dari atas ke bawah.
4. **Multi-If Terpisah**: Rangkaian statement `if` yang ditulis secara independen (tidak saling terkait dengan `elif`). Cocok untuk skenario di mana banyak kondisi dapat terpenuhi secara bersamaan (misalnya akumulasi diskon).
5. **Match-Case**: Fitur *Structural Pattern Matching* yang hadir mulai dari Python 3.10. Sangat efektif menggantikan rangkaian `if-elif` panjang saat mencocokkan satu nilai variabel terhadap sejumlah konstanta pasti.
6. **Operator Logika**: Operator seperti `and`, `or`, dan `not` yang digabungkan ke dalam percabangan untuk membentuk kondisi *majemuk*.

## 2.2 Kode Program

### 2.2.1 Program "if1_2611533010.py"


```python
# Buat file dengan nama if1_2611533010.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit NIM terakhir contoh: ipk_3010
# program ini menggunakan fungsi input()
ipk_3010 = float(input("Input IPK Anda = "))

if ipk_3010 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3010))

print("Program Selesai")
```

**Output :**



**Penjelasan :**
Program ini mengimplementasikan logika **If Tunggal**. Program akan meminta konversi input IPK tipe *float*, dan blok pencetakan status kelulusan hanya akan tereksekusi apabila kondisi (`ipk_3010 > 2.75`) terpenuhi. Jika tidak memenuhi kriteria, program langsung melompat ke baris pencetakan penyelesaian program tanpa ada jalan alternatif lain.

### 2.2.2 Program "if2_2611533010.py"


```python
# Buat file dengan nama if2_2611533010.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit NIM terakhir contoh: ipk_3010
# program ini menggunakan fungsi input()

ipk_3010 = float(input("Input IPK Anda = "))

if ipk_3010 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3010))
else:
    print("Anda tidak lulus")
print("Program Selesai")
```

**Outpuut**




**Penjelasan program if-else di atas:**

Kode ini mengembangkan logika latihan pertama dengan menyuntikkan konsep jalan bercabang ganda **If-Else**. Apabila hasil evaluasi ekspresi kondisi utama salah (kondisi jatuh menjadi *False* akibat IPK <= 2.75), maka komputasi tidak akan langsung mengakhiri blok percabangan, melainkan wajib mengeksekusi parameter penyangga yang diletakkan di dalam badan `else`.

### 2.2.3 Program "multi_if1_2611533010.py"


```python
# Buat file dengan nama multi_if1_2611533010.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit NIM terakhir contoh: ipk_3010
# program ini menggunakan fungsi input()

umur_3010 = int(input("Input Umur Anda: "))
sim_3010 = input("Apakah Anda sudah memiliki SIM C? (y/t): ")[0]

if umur_3010 >= 17 and sim_3010 == "y":
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_3010 >= 17 and sim_3010 != "y":
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")

if umur_3010 < 17 and sim_3010 == "y":
    print("Anda belum cukup umur punya SIM")

if umur_3010 < 17 and sim_3010 != "y":
    print("Anda belum cukup umur bawa motor")
```

**Output :**




**Penjelasan program multi-if di atas:**

Program ini adalah bentuk dari blok **Multi-If** tunggal yang berdiri berurutan. Masing-masing kondisi dianalisis menggunakan operator logika `and`. Meskipun alur kodenya berhasil menampilkan peringatan kendaraan, namun teknik evaluasi penulisan seperti ini secara komputasi tidak terlalu optimal karena interpreter akan memeriksa keempat pernyataan *if* secara paksa, terlepas apakah kondisi pertama sudah terpenuhi atau belum.

### 2.2.4 Program "if_elif_else1_2611533010.py"


```python
# Buat file dengan nama if_elif_else1_2611533010.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit NIM terakhir contoh: ipk_3010
# program ini menggunakan fungsi input()

umur_3010 = int(input("Input Umur Anda: "))
sim_3010 = input("Apakah Anda sudah memiliki SIM C? (y/t): ")[0]

if umur_3010 >= 17 and sim_3010 == "y":
    print("Anda sudah dewasa dan boleh bawa motor")
elif umur_3010 >= 17 and sim_3010 != "y":
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_3010 < 17 and sim_3010 == "y":
    print("Anda belum cukup umur punya SIM")
else:
    print("Anda belum cukup umur dan tidak boleh bawa motor")
print("Program Selesai")
```

**Output :**


**Penjelasan :**
Modifikasi mutlak dari program "multi_if1_2611533010.py", namun kali ini algoritma berjalan jauh lebih ideal melalui konsep penulisan **If-Elif-Else** utuh. Dengan konstruksi ini, siklus pemeriksaan syarat tidak saling tumpang-tindih. Apabila komparator logika sudah terpenuhi di `elif` urutan pertama, program segera melepaskan blok lain sehingga memori tidak sia-sia melakukan pengecekan berulang.

### 2.2.5 Program "multi_if2_2611533010.py"


```python
# Buat file dengan nama multi_if2_2611533010.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit NIM terakhir contoh: total_belanja_3010
# program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_3010 = float(input("Input Total Belanja Anda (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3010 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member_3010 = input_member_3010 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3010 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3010 = input_promo_3010 in ["y", "ya"]

total_diskon_persen_3010 = 0

# Multi-IF terpisah: setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3010 > 1000000:
    total_diskon_persen_3010 += 10 # Diskon belanja besar

if is_member_3010:
    total_diskon_persen_3010 += 5 # Diskon member

if kode_promo_valid_3010:
    total_diskon_persen_3010 += 15 # Diskon voucher

# menghitung nominal diskon dan total bayar
nominal_diskon_3010 = total_belanja_3010 * (total_diskon_persen_3010 / 100)
total_bayar_3010 = total_belanja_3010 - nominal_diskon_3010

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_3010}% (Rp {nominal_diskon_3010:,.0f})")
print(f"Total Bayar  : Rp {total_bayar_3010:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_3010}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Output :**



**Penjelasan :**
Skenario spesifik di mana kita wajib memakai blok percabangan diskrit / **Multi-If Terpisah**. Kita dilarang memakai `elif` di sini karena tujuan program adalah memfasilitasi terjadinya akumulasi atau "penumpukan" diskon. Masing-masing baris `if` dieksekusi secara independen dan menambahkan persentase nilainya secara bertingkat apabila parameter pengguna (member, promo, total belanja) tervalidasi semuanya secara simultan.

### 2.2.6 Program "nilai_2611533010.py"


```python
# Buat file dengan nama nilai_2611533010.py
# Buat program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_3010
# Program ini menggunakan fungsi input()
# program konversi nilai angka menjadi huruf

nilai_3010 = int(input("Inputkan nilai angka= "))

if nilai_3010 >= 81:
    print("A")
elif nilai_3010 >= 70:
    print("B")
elif nilai_3010 >= 60:
    print("C")
elif nilai_3010 >= 50:
    print("D")
else:
    print("E")
```

**Output :**


**Penjelasan :**
Program ini merupakan bentuk penerapan **If-Elif-Else** secara berjenjang untuk menangani evaluasi dengan lebih dari 2 percabangan logika. Program menyeleksi nilai angka secara sekuensial (mulai dari rentang nilai tertinggi `81`). Penggunaan `elif` memastikan apabila satu kondisi *grade* sudah terpenuhi, maka *grade* di bawahnya secara otomatis akan diabaikan.

### 2.2.7 Program "match_case1_2611533010.py"


```python
# Buat file dengan nama match_case1_2611533010.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_3010
# Program ini menggunakan fungsi input()
# program konversi angka menjadi nama bulan
bulan_3010 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_3010:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Output :**



**Penjelasan :**
Fitur **Match-Case** digunakan sebagai solusi efisien untuk percabangan struktural spesifik yang melakukan perbandingan nilai eksak/langsung (bukan operasi komparasi rentang < >). Pengguna memasukkan variabel bertipe *integer*, lalu alur perbandingan lompat ke nilai `case` yang identik. Penggunaan blok `case _` berfungsi layaknya `else`, di mana ia menangani seluruh probabilitas input di luar opsi 1-12.

## Tugas Praktikum

**Soal :**

Studi Kasus: Sistem Loket Terpadu & Audit Transaksi Ekspedisi Wahana
Sebuah wahana rekreasi petualangan membutuhkan sistem loket berbasis konsol untuk menentukan jenis wahana, memverifikasi kelaikan pengunjung dengan kondisi majemuk (logical operators), menghitung diskon akumulatif menggunakan multi-if terpisah, serta mencetak audit status operasional.
Catatan Materi: Seluruh alur program disusun murni menggunakan 4 pilar percabangan yang diajarkan pada modul Pekan 4: (1) if tunggal, (2) if-else, (3) if-elif-else dengan operator logika (and, or, !=), (4) Multi-if terpisah untuk diskon akumulasi, dan (5) match-case (1 s.d. 5 beserta case _). Tidak ada nested-if (if di dalam if).

**Kode Jawaban :**

a. Kode Program "Tugas4_2611533010.py"

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# 1. INPUT DATA PENGUNJUNG
nama_pengunjung_3010 = input("Masukkan Nama Pengunjung        : ")
umur_3010 = int(input("Input umur anda                 : "))
sim_3010 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()
jumlah_tiket_3010 = int(input("Masukkan jumlah tiket           : "))

# Validasi jumlah tiket menggunakan if tunggal
if jumlah_tiket_3010 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")
    raise SystemExit


# 2. PEMILIHAN PAKET WAHANA MENGGUNAKAN MATCH-CASE
print("\nPilihan Paket Wahana (1-5):")
print(" 1. Safari Rimba         (Rp 50,000)")
print(" 2. Arung Jeram          (Rp 75,000)")
print(" 3. Motor ATV Ekstrim    (Rp 120,000)")
print(" 4. Roller Coaster Kilat (Rp 100,000)")
print(" 5. All-Access VIP       (Rp 220,000)")

paket_3010 = int(input("Masukkan nomor paket (1-5) : "))

match paket_3010:
    case 1:
        nama_wahana_3010 = "Wahana Safari Rimba"
        harga_satuan_3010 = 50000
    case 2:
        nama_wahana_3010 = "Wahana Arung Jeram"
        harga_satuan_3010 = 75000
    case 3:
        nama_wahana_3010 = "Wahana Motor ATV Ekstrim"
        harga_satuan_3010 = 120000
    case 4:
        nama_wahana_3010 = "Wahana Roller Coaster Kilat"
        harga_satuan_3010 = 100000
    case 5:
        nama_wahana_3010 = "Wahana All-Access VIP"
        harga_satuan_3010 = 220000
    case _:
        print("Paket wahana tidak valid!")
        raise SystemExit


# 3. VALIDASI IZIN KENDALI WAHANA
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_3010 == 3 and umur_3010 >= 17 and sim_3010 == 'y':
    status_akses_3010 = "Anda sudah dewasa dan boleh mengendarai ATV sendiri."
elif paket_3010 == 3 and umur_3010 >= 17 and sim_3010 != 'y':
    status_akses_3010 = "Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur)."
elif paket_3010 == 3 and umur_3010 < 17 and sim_3010 == 'y':
    status_akses_3010 = "Identitas tidak valid: Belum cukup umur memiliki SIM."
elif paket_3010 == 3:
    status_akses_3010 = "Anda belum cukup umur dan tidak boleh bawa motor ATV."
elif umur_3010 >= 10:
    status_akses_3010 = "Anda memenuhi syarat umur untuk wahana ini."
else:
    status_akses_3010 = "Anda belum cukup umur untuk wahana ini."

print(f"Status Akses: {status_akses_3010}")


# 4. INPUT DATA DISKON
is_member_3010 = input("Apakah Anda member? (y/t)       : ").strip().lower()
kode_promo_valid_3010 = input("Apakah kode promo valid? (y/t) : ").strip().lower()


# 5. PERHITUNGAN SUBTOTAL
subtotal_3010 = harga_satuan_3010 * jumlah_tiket_3010


# 6. MULTI-IF UNTUK DISKON AKUMULATIF
total_diskon_persen_3010 = 0

if subtotal_3010 >= 200000:
    total_diskon_persen_3010 += 10

if is_member_3010 in ['y', 'ya']:
    total_diskon_persen_3010 += 5

if kode_promo_valid_3010 in ['y', 'ya']:
    total_diskon_persen_3010 += 15

if jumlah_tiket_3010 >= 5:
    total_diskon_persen_3010 += 5


# 7. PERHITUNGAN NOMINAL DISKON DAN TOTAL BAYAR
nominal_diskon_3010 = subtotal_3010 * (total_diskon_persen_3010 / 100)
total_bayar_3010 = subtotal_3010 - nominal_diskon_3010


# 8. EVALUASI KELULUSAN AUDIT
if total_bayar_3010 > 300000:
    catatan_layanan_3010 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_3010 = "Terima kasih telah berkunjung."


# 9. RINCIAN PEMBAYARAN
print("\n--- Rincian Pembayaran ---")
print(f"Nama Pengunjung  : {nama_pengunjung_3010}")
print(f"Wahana           : {nama_wahana_3010}")
print(f"Harga Satuan     : Rp {harga_satuan_3010:,.0f}".replace(",", "."))
print(f"Jumlah Tiket     : {jumlah_tiket_3010}")
print(f"Subtotal Belanja : Rp {subtotal_3010:,.0f}".replace(",", "."))
print(f"Total Diskon     : {total_diskon_persen_3010}% (Rp {nominal_diskon_3010:,.0f})".replace(",", "."))
print(f"Total Bayar      : Rp {total_bayar_3010:,.0f}".replace(",", "."))
print(f"Catatan Layanan  : {catatan_layanan_3010}")

print("\nProgram Selesai")
```

**Output :**


**Penjelasan :**

Program Sistem Loket Terpadu ini dirancang secara khusus untuk mematuhi aturan dasar penggunaan 4 pilar percabangan Python tanpa memanfaatkan nested-if (if bersarang) maupun penghentian paksa sistem. Seluruh alur program dijalankan secara linear dan sistematis melalui beberapa poin utama berikut:
1. Penerapan Logika Saklar Sakral (Boolean Flag)
Untuk menghindari struktur nested-if, program menggunakan sebuah variabel kontrol (misalnya validasi_3010) yang bernilai awal True. Variabel ini bertindak layaknya "saklar utama". Jika ditemukan kesalahan input seperti kuota tiket bernilai nol/minus atau nomor paket yang salah, saklar ini otomatis berubah menjadi False, sehingga blok pemrosesan selanjutnya dapat dilewati secara aman.
2. Pilar 1: Validation Check dengan If Tunggal
Blok awal menggunakan struktur if tunggal untuk memverifikasi kelogisan kuota tiket yang dimasukkan. Apabila nilai kuota \le 0, sistem secara langsung memberikan instruksi peringatan dan mengubah status saklar validasi menjadi False.
3. Pilar 2: Pemetaan Paket Menggunakan Match-Case
Fitur Pattern Matching (match-case) dipakai untuk mencocokkan angka pilihan wahana dari pengguna:
 * case 1 sampai case 5: Mengalokasikan deskripsi nama wahana serta besaran harga tiket secara presisi.
 * case _: Berperan sebagai penangan default jika nomor wahana tidak tersedia di menu, lalu mengubah nilai saklar validasi menjadi False.
4. Pilar 3: Validasi Akses Kendali dengan If-Elif-Else Terstruktur
Tahap ini memeriksa kriteria batas usia dan kepemilikan SIM C pengunjung. Seluruh syarat dikombinasikan secara datar menggunakan operator logika (and, !=) dalam satu deretan if-elif-else tanpa adanya if di dalam if. Penilaian dilakukan berurutan mulai dari batas syarat paling ketat (seperti wahana ATV Ekstrim) hingga syarat umum untuk wahana lainnya.
5. Pilar 4: Kalkulasi Diskon Menggunakan Multi-If Terpisah
Penentuan potongan harga sengaja tidak menggunakan elif agar sifat perhitungannya bisa saling menambah (akumulatif). Setiap kriteria diskon—mulai dari total belanja \ge \text{Rp } 200.000, status keanggotaan member, klaim kode promo, hingga kuota tiket rombongan (\ge 5)—diejawantahkan dalam blok if independen sehingga besaran persen diskon dapat terus ditumpuk secara dinamis.
6. Pilar 5: Evaluasi Audit dan Format Output (If-Else)
Pada tahap akhir, blok if-else memverifikasi total tagihan bersih untuk menentukan hak kualifikasi hadiah souvenir (apabila total bayar > \text{Rp } 300.000). Seluruh angka finansial juga disajikan secara rapi menggunakan f-string formatting (:,.0f) untuk menyisipkan tanda titik sebagai pemisah ribuan otomatis khas mata uang Rupiah.

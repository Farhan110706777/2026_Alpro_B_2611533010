## 2.1 Dasar Teori
1. Variabel dan Dynamic Typing: Di dalam Python, variabel tidak perlu dideklarasikan tipe datanya secara eksplisit. Python otomatis mendeteksi tipe data berdasarkan nilai yang diisikan ke dalam variabel tersebut.
2. Kategori Tipe Data Primitif:

A.) Numerik: Digunakan untuk perhitungan matematika. Terdiri dari int (bilangan bulat), float (bilangan desimal), dan complex (bilangan imajiner/kompleks)
.
B.) String (str): Kumpulan karakter teks yang diapit oleh tanda petik tunggal, ganda, ataupun tiga pasang petik (multiline).

C.) Boolean (bool): Tipe data yang hanya menyimpan dua nilai logika, yaitu True (benar) atau False (salah).

D.) Input data dinamis (input()): Fungsi bawaan untuk menerima ketikan dari pengguna melalui layar console. Seluruh input yang masuk akan dianggap bertipe string secara default.

E.) Konstanta: Konsep penetapan variabel yang nilainya bersifat konstan atau ditujukan agar tidak diubah-ubah sepanjang program berjalan, yang dalam Python umumnya dipertegas menggunakan penanda Final dari modul typing

## 2.2 Kode Program

### 2.2.1 Program "numerik_2611533010.py"
Program berikut mengilustrasikan penciptaan beberapa variabel, masing-masing dengan muatan tipe numerik yang berbeda (integer, float, complex), kemudian mengidentifikasi validasi struktur bawaannya memakai built-in function `type()`.

```python
# Buat file dengan nama numerik_2611533010.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_3010
# creat a variabel with integer value.
a_3010 = 100
print("The type of variable having value", a_3010, "is", type(a_3010))

# creat a variabel with float value
b_3010 = 10.2345
print("The type of variable having value", b_3010, "is ", type(b_3010))

# creat a variabel with complex value.
c_3010 = 100 + 3j
print("The type of variable having value", c_3010, " is ", type(c_3010))
```

**Penjelasan :**
1. `a_3010 = 100` : Berfungsi untuk mendeklarasikan variabel `a_3010` dan menginisialisasinya dengan nilai bilangan bulat (integer), yaitu `100`.
2. `b_3010 = 10.23455` : Berfungsi untuk mendeklarasikan variabel `b_3010` dengan menugaskan nilai berupa pecahan atau desimal (float).
3. `c_3010 = 100+3j` : Berfungsi untuk mendeklarasikan variabel `c_3010` dengan nilai bilangan kompleks (complex) bawaan Python yang ditandai dengan sufiks `j`.
4. `type(...)` pada fungsi `print` : Berfungsi sebagai *built-in function* dari Python yang digunakan untuk mengecek tipe data aktual dari variabel yang ada di dalam tanda kurungnya, dan mencetak nama kelas tipe datanya ke layar terminal.

### 2.2.2 Program String "string_2611533010"
Blok program identitas diri di bawah ini mengandalkan string untuk mendefinisikan informasi profil. Teks dapat ditampung dalam sebaris tanda petik maupun kutipan multiline string ketika mendefinisikan lokasi alamat secara berderet.

```python
# Buat file dengan nama string_2611533010.py
#nama variabel ditambah 4 digit nim terakhir contoh: nama_3010
#ganti programnya dan data masing2
nama_3010 = "Muhammad Farhan Amri"
jenis_kelamin_3010 = 'L'
alamat_3010 = """
Lubuk Lintah
kec.kuranji
kota Padang
"""
agama_3010 = 'Islam'

print("Nama saya adalah: ", nama_3010)
print("Jenis kelamin saya adalah: ", jenis_kelamin_3010)
print("Alamat saya adalah: ", alamat_3010)
print("Agama saya adalah:", agama_3010)
```

Penjelasan program string di atas:

1. `nama_3010 = "Muhammad Farhan Amri"` dan `jenis_kelamin_3010 = 'L'` : Berfungsi menginisiasi variabel tipe data teks (string). Python mengizinkan pemakaian tanda kutip dua (`" "`) maupun tanda kutip tunggal (`' '`) untuk membungkus sebaris teks.
2. `alamat_3010 = """..."""` : Berfungsi untuk menampung teks yang berisi lebih dari satu baris atau *multiline string*. Teks dibungkus menggunakan tanda kutip tiga (bisa `"""` atau `'''`) agar perpindahan baris (enter) ikut tersimpan sebagai bagian dari data alamat tersebut.
3. `print("Nama saya adalah:", nama_3010)` : Berfungsi untuk mencetak gabungan string karakter statis dengan nilai yang tersimpan di dalam suatu variabel string.

### 2.2.3 Program Boolean "boolean_2611533010"
Program berikut mengaplikasikan boolean untuk mengecek syarat parameter nilai kelulusan. Status nilai akan dicari kepastiannya (`True` atau `False`) sebelum divalidasi dan diizinkan melaju memanggil perintah cetak Cum Laude.

```python
# Buat file dengan nama 2611533010.py
# Nama variabel ditambah 4 digit nim terakhir contoh: nama_3010
# Deklarasi variabel dengan tipe data boolean

is_lulus_3010 = True
is_cumlaude_3010 = True

# Menggunakan Boolean
nilai_3010 = 85
batas_lulus_3010 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_3010 = nilai_3010 >= batas_lulus_3010  # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_3010)
print("Apakah Lulus:", status_kelulusan_3010)
if is_lulus_3010 and is_cumlaude_3010:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

Penjelasan program boolean di atas:

1. `is_lulus_3010 = True` dan `is_cumlaude_3010 = True` : Berfungsi mendeklarasikan variabel tipe data Boolean yang merepresentasikan kebenaran, di mana nilainya hanya dapat diatur sebagai `True` (benar) atau `False` (salah).
2. `status_kelulusan_3010 = nilai_3010 >= batas_lulus_3010` : Berfungsi melakukan evaluasi operator perbandingan relasional (lebih besar sama dengan). Karena nilai 85 lebih besar dari 75, hasil evaluasi tersebut akan mengembalikan nilai kebenaran Boolean berupa `True` lalu disalurkan ke variabel `status_kelulusan_3010`.
3. `if is_lulus_3010 and is_cumlaude_3010:` : Berfungsi sebagai blok pengkondisian menggunakan operator logika `and`. Program hanya akan mengeksekusi fungsi print berupa ucapan "Selamat!" yang ada di bawahnya apabila kondisi di kiri dan kanan sintaks `and` sama-sama bernilai `True`.

### 2.2.4 Program Input "input_2611533010"
Skrip program ini bergantung pada `input()` agar sisi pengerjaan perumusan dapat secara fleksibel disuplai masukan dari sisi end-user. Masukan awal direkam menggunakan string default untuk lalu dipaksa berubah bentuk (`type casting`) menjadi bilangan `int` agar kalkulasi dimensi berjalan lancar.

```python
# Buat file dengan nama input_2611533010.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: nama_3010
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_3010 = input('Masukkan nilai panjang: ')
lebar_3010 = input('Masukkan nilai lebar: ')
luas_3010 = int(panjang_3010) * int(lebar_3010)
print("Luas:", luas_3010)
```

![Output Input](Screenshot%202026-09-15%20225542.png)



**Penjelasan :**
1. `input('Masukkan nilai panjang: ')` : Berfungsi menampilkan teks sapaan ke layar, kemudian menahan eksekusi program sejenak untuk menunggu pengguna mengetikkan sesuatu (inputan) ke dalam *console* / *terminal*. Secara bawaan (default), semua masukan akan direkam sebagai data String.
2. `int(panjang_3010)` dan `int(lebar_3010)` : Berfungsi melakukan proses *type casting* atau pemaksaan konversi tipe data.[cite: 2] Nilai panjang dan lebar yang awalnya berupa karakter teks atau String diubah paksa menjadi bilangan bulat (Integer) agar dapat dikalikan dalam operasi matematika `*` (perkalian).
3. `luas_3010 = ...` : Menyimpan hasil akhir nilai perkalian yang bertipe Integer.

### 2.2.5 Program Konstanta "konstanta_2611533010"
Menggunakan import atribut `Final` dari modul `typing`, kode ini mempatenkan variabel global `PI` senilai 3.14. Tujuannya adalah mereduksi kesalahan perubahan nilai sembarang secara tidak sengaja ketika program dikomputasi untuk menelusuri luas lingkaran.

```python
# Buat file dengan nama konstanta_2611533010.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran.
# Nama variabel ditambah 4 digit nim terakhir contoh: jari_3010

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_3010 = float(input("Masukkan nilai jari-jari: "))
luas_3010 = PI * jari_3010 * jari_3010
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3010, luas_3010))
```

![Output Konstanta](output_konstanta.png)

**Penjelasan :**
1. `from typing import Final` : Berfungsi untuk memanggil library atau modul bawaan `typing` dari Python, guna menggunakan kemampuan pelabelan (anotasi) jenis variabel tertentu dalam hal ini sebagai tipe `Final`.
2. `PI: Final = 3.14` : Mendeklarasikan bahwa variabel `PI` bersifat final atau konstan. Ini menunjukkan pada struktur kode bahwa `PI` merupakan tetapan nilai matematis yang tidak boleh diubah oleh baris kode mana pun secara tidak sengaja di masa mendatang.
3. `float(input(...))` : Meminta input dari pengguna, namun berkat pembungkusan fungsi tipe data `float(...)`, teks angka yang dimasukkan pengguna secara sekejap langsung dikonversi ke tipe bilangan desimal (float).
4. `%f` dan `%.2f` : Berfungsi sebagai teknik pemformatan string kuno gaya bahasa C (*String Formatting*). Simbol `%f` mencetak bilangan float secara utuh, sedangkan `%.2f` digunakan untuk membatasi tampilan hasil hitung luas dan input jari-jari agar hanya menyisakan dua digit angka spesifik saja setelah tanda koma.

Rancang sebuah program konsol sederhana yang mencatat data identitas praktikan baru 
dan melakukan verifikasi status kelulusan awal praktikan secara otomatis dengan syarat:

1. Nama berkas Tugas2_2611533004.py dan seluruh variabel diakhiri dengan 4 digit terakhir NIM (...3004).
2. Terapkan konstanta batas kelulusan menggunakan modul typing.Final.
3. Terapkan tipe data String & Char (untuk nama, jenis kelamin, dan multiline string untuk alamat).
4. Terapkan tipe data Numerik & Type Casting (int untuk umur, float untuk skor, dan complex untuk token identifikasi).
5. Evaluasi kelulusan secara dinamis menjadi tipe data Boolean dan tampilkan seluruh tipe data ke layar menggunakan fungsi bawaan type().

```python
Kode Jawaban
from typing import Final

# 1. Penerapan Konstanta
BATAS_LULUS: Final = 75.0

# Header Program
print("=== SISTEM REGISTRASI PRAKTIKAN ALPRO 2026 ===")

# 2. Input Data & Type Casting
nama_3010 = input("Masukkan Nama Mahasiswa   : ")
jk_3010 = input("Masukkan Jenis Kelamin (L/P): ")
umur_3010 = int(input("Masukkan Umur            : "))
skor_3010 = float(input("Masukkan Skor Tes Awal   : "))

# Alamat multiline
alamat_3010 = """Jl. Muhammad Yunus No 4,
Kecamatan Kuranji,
Kota Padang"""

# Token Sinyal (Complex)
token_3010 = 100 + 3j

# Evaluasi Status Kelulusan (Bool)
status_lulus_3010 = skor_3010 >= BATAS_LULUS

# 3. Output Data & Tipe Data
print("\n=== DATA PRAKTIKAN & HASIL PEMERIKSAAN ===")
print(f"Nama Mahasiswa : {nama_3010} | Tipe: {type(nama_3010)}")
print(f"Jenis Kelamin  : {jk_3010} | Tipe: {type(jk_3010)}")
print(f"Alamat Domisili:\n{alamat_3010}\n| Tipe: {type(alamat_3010)}")
print(f"Umur           : {umur_3010} tahun | Tipe: {type(umur_3010)}")
print(f"Skor Tes Awal  : {skor_3010} | Tipe: {type(skor_3010)}")
print(f"ID Token Sinyal: {token_3010} | Tipe: {type(token_3010)}")

print("\n=== STATUS KELULUSAN PRAKTIKUM ===")
print(f"Batas Minimum Nilai: {BATAS_LULUS}")
print(f"Apakah Dinyatakan Lulus?: {status_lulus_3010} | Tipe: {type(status_lulus_3010)}")
```
